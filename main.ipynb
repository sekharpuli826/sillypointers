```python
print("🏏 Cricket Tournament Manager")
print("1. Set Tournament Name")
print("2. Add Team")
print("3. List Teams")
print("4. Add Player")
print("5. List Players")
print("6. Create Match")
print("7. List Matches")
```

```python
choice = input("Choose an option: ")

if choice == "1":
    name = input("Tournament name: ")
    conn = sqlite3.connect("cricket.db")
    cur = conn.cursor()
    cur.execute("DELETE FROM tournament")  # keep only one name
    cur.execute("INSERT INTO tournament (name) VALUES (?)", (name,))
    conn.commit()
    conn.close()
    print("Tournament named:", name)

elif choice == "2":
    name = input("Team name: ")
    add_team(name)
    print("Team added:", name)

elif choice == "3":
    for t in list_teams():
        print(f"ID={t[0]}  Name={t[1]}")

elif choice == "4":
    team_id = int(input("Team ID: "))
    name = input("Player name: ")
    role = input("Role (batsman/bowler/all-rounder/keeper): ")
    pos = int(input("Batting position (1=opener, 3=one-down): "))
    add_player(name, role, team_id, pos)
    print("Player added:", name)

elif choice == "5":
    team_id_raw = input("Team ID (blank for all): ").strip()
    team_id = int(team_id_raw) if team_id_raw else None
    for p in list_players(team_id):
        print(f"ID={p[0]}  Name={p[1]}  Role={p[2]}  Pos={p[3]}")

elif choice == "6":
    team1_id = int(input("Team 1 ID: "))
    team2_id = int(input("Team 2 ID: "))
    overs = int(input("Overs: "))
    venue = input("Venue: ")
    date = input("Date (YYYY-MM-DD): ")
    create_match(team1_id, team2_id, overs, venue, date)
    print("Match created.")

elif choice == "7":
    for m in list_matches():
        print(m)
```
